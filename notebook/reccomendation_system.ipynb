```python
## Flattening dictionary to string
```

```python
import ast
def removeDictionary(dictString: str):
    try:
        dict_list = ast.literal_eval(dictString)
        lists = []
        for i in dict_list:
            lists.append(i["name"].replace(" ", ""))
        return " ".join(lists)
    except:
        return dictString
```

```python
movies_df[['genres', 'keywords', 'production_companies', 'spoken_languages']] = movies_df[['genres', 'keywords', 'production_companies', 'spoken_languages']].map(removeDictionary)
```

```python
credits_df[['cast', 'crew']] = credits_df[['cast', 'crew']].map(removeDictionary)
```

## Reccomendatoin System

```python
import nltk
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
counts = CountVectorizer(stop_words='english', max_features=10000)
x = counts.fit_transform(df['tags']).toarray()
```

```python
similarity = cosine_similarity(x)
```

```python
def recommended_movies(index: int):
    try:
        distances = similarity[index]
        movie_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x: x[1])[1:11]
        print(f"Recommendations for '{df.iloc[index].title}':")
        for i in movie_list:
            print(df.iloc[i[0]].title)
    except IndexError:
        print("Movie not found in dataset.")
```
